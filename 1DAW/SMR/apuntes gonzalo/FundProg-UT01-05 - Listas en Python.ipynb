# Listas

### **1. ¿Qué son las listas en Python?**

Las listas son un tipo de dato complejo y particular del lenguaje de programación Python. Una lista se corresponde con una colección arbitraria de objetos. Las listas son similares a estructuras como los arrays en otros lenguajes de programación con la diferencia de que aportan una mayor flexibilidad.

Las listas en Python se representan con el tipo `list` y la sintaxis que se utiliza para definirlas consiste en indicar una lista de objetos separados entre comas y encerrados entre corchetes: `[obj1, obj2, ..., objn]`

````python
lista1 = [1, 2, 3, 4, 5]
print(type(lista1))
print(lista1)
````

### **2. Las listas respetan el orden**

Una de las características importantes de las listas es que se corresponden con una colección ordenada de objetos. El orden en el que se especifican los elementos cuando se define una lista es relevante y se mantiene durante toda su vida.

**Podemos utilizar los operadores como (in, ==) para comparar listas**

````python
lista1 = [1, 2, 3, 4, 5]
lista2 = [5, 4, 3, 2, 1]
print(lista1 == lista2)
````

````python
lista1 = [1, 2, 3, 4, 5]
lista2 = [1, 2, 3, 4, 5]
print(lista1 == lista2)
````

### **3. Las listas pueden contener objetos de diferentes tipos**

Las listas pueden contener una colección de objetos de diferentes tipos.

````python
lista1 = [1, "hola", 3.14, True]
print(lista1)
````

````python
lista1 = [1, "hola", 3.14, ["Gonzalo", 16]]
print(type(lista1))
print(lista1)
````

Una cosa interesante es que una lista puede llegar a contener una función.

````python
lista1 = [1, "hola", 3.14, True, len("hola")]
print(lista1)
````

Los elementos de una lista no tienen que ser únicos. Puede repetirse el mismo elemento varias veces en la misma lista

````python
lista1 = [1, "hola", 3.14, "hola", 1, "hola", 3.14]
print(lista1)
````

### **4. Acceso a los elementos de una lista**

Los elementos de una lista pueden ser accedidos utilizando el concepto y la sintaxis de la indexación que se presentó cuando hablamos de strings.

Todas las variaciones vistas en la sección de strings (indexing, slicing, stride) aplican a las listas.

#### **Indexing**

````python
lista1 = [1, "hola", 3.14, "hola", 1]
print(type(lista1))
print(lista1)
print(lista1[0])
print(lista1[1])
print(lista1[2])
print(lista1[3])
print(lista1[4])
````

#### **Slicing**

````python
lista1 = [1, "hola", 3.14, "hola", 1]
print(type(lista1))
print(lista1)
print(lista1[0:2])
print(lista1[1:3])
print(lista1[2:4])
````

````python
lista1 = [1, "hola", 3.14, "hola", 1]
print(type(lista1))
print(lista1)
print(lista1[:2])
print(lista1[1:])
print(lista1[:])
````

#### **Stride**

````python
lista1 = [1, "hola", 3.14, "hola", 1]
print(type(lista1))
print(lista1)
print(lista1[::2])
print(lista1[1::2])
````

Una cosa interesante es que podemos utilizar el concepto de stride para darle la vuelta a una lista

````python
lista1 = [1, "hola", 3.14, "adios", 8]
print(type(lista1))
print(lista1)
print(lista1[::-1])
````

### **5. Operaciones con listas**

Como comentaba anteriormente, las listas soportan muchos de los operadores y funciones de Python por defecto, veamos cómo trabajar con operaciones con dos listas.(**\*, +, len, min, max**)

````python
lista1 = [1, 2, 3]
lista2 = [4, 5, 6]
print(lista1 + lista2)
````

````python
lista1 = [1, 2, 3]
print(lista1 * 3)
````

````python
lista1 = [1, 2, 3]
print(len(lista1))
````

````python
lista1 = [10, 1, 2, 3]
print(min(lista1))
````

````python
lista1 = [1, 2, 10, 3]
print(max(lista1))
````

### **6. Las listas se pueden anidar**

Tal y como se ha presentado anteriormente, una lista puede contener cualquier tipo de objeto. Esto incluye otra lista. Una lista puede contener sublistas, que a su vez pueden contener sublistas, y así hasta una profundidad arbitraria.

````python
lista1 = [1, 2, [80, 90]]
print(lista1)
````

````python
lista1 = [1, 2, 3]
lista2 = [4, 5, 6]
lista3 = [lista1, lista2]
print(lista3)
````

````python
lista1 = [1, 2, [10, [50, 60]]]
print(lista1)
````

````python
lista1 = [1, 2, 3]
lista2 = [4, 5, 6]
lista3 = [lista1, lista2]
print(lista3)
print(lista3[0])
print(lista3[1])
````

Para aceder a los elementos de una sublista utilizamos la sintaxis `[][]`

````python
lista1 = [1, 2, 3]
lista2 = [4, 5, 6]
lista3 = [lista1, lista2]
print(lista3[0][0])
print(lista3[0][1])
````

````python
lista1 = [1, 2, [10, [50, 60]]]
print(lista1[2][1][0])
print(lista1[2][1][1])
````

No hay límite en la cantidad de listas que podemos anidar, pueden ser tantas como soporte la memoria de nuestro sistema.

Es importante entender que los operadores como (**in**) se aplicarán sobre la primera lista y no aplicarán de manera recursiva a las sublistas que estos contuvieran.

````python
lista1 = [1, 2, 3]
print(2 in lista1)
````

````python
lista1 = [1, 2, 3]
lista2 = [4, 5, 6]
lista3 = [lista1, lista2]
print(lista3)
print(2 in lista3)
````

````python
lista1 = [1, 2, [10, [50, 60]]]
print(60 in lista1)
````

### **7. Las listas son mutables**

Al contrario que lo que veíamos cuando hablabamos de strings, las listas son estructuras mutables. Esto quiere decir que podemos modificar los elementos que forman parte de ellas.

````python
lista1 = [1, 2, 3]
lista1[0] = 10
print(lista1)
````

````python
lista1 = [1, 2, 3]
lista1[1] = 20
print(lista1)
````

````python
lista1 = [1, 2, 3]
lista1[2] = 30
print(lista1)
````

````python
lista1 = [1, 2, 3]
lista1[0] = 10
lista1[1] = 20
lista1[2] = 30
print(lista1)
````

````python
lista1 = [[1, 2, 3], [20, 40, 60]]
lista1[0][0] = 10
lista1[1][1] = 90
print(lista1)
````

También podemos eliminar elementos de una lista usando el comando **del** seguido del elemento de la lista a eliminar

````python
lista1 = [1, 2, 3]
del lista1[0]
print(lista1)
````

````python
lista1 = [1, 2, 3]
del lista1[1]
print(lista1)
````

````python
lista1 = [1, 2, 3]
del lista1[2]
print(lista1)
````

````python
lista1 = [1, 2, 3]
del lista1[0]
del lista1[0]
print(lista1)
````

````python
lista1 = [1, 2, 3]
del lista1[0]
del lista1[0]
del lista1[0]
print(lista1)
````

Podemos utilizar otros conceptos de indexación, como el slicing o el stride para modificar varios elementos de una lista

````python
lista1 = [1, 2, 3]
lista1[0:2] = [10, 20]
print(lista1)
````

````python
lista1 = [1, 2, 3]
lista1[::2] = [10, 20]
print(lista1)
````

El número de elementos seleccionados no tiene que ser igual a los asignados

````python
lista1 = [1, 2, 3]
lista1[0:2] = [10]
print(lista1)
````

````python
lista1 = [1, 2, 3]
lista1[0:2] = [10, 20, 30]
print(lista1)
````

````python
lista1 = [1, 2, 3]
lista1[::2] = [10, 20]
print(lista1)
````

### **8. Métodos de Listas**

#### **8.1. `append(elemento)`**
Añade un elemento al final de la lista.

````python
lista1 = [1, 2, 3]
lista1.append(4)
print(lista1)
````

````python
lista1 = [1, 2, 3]
lista1.append([4, 5, 6])
print(lista1)
````

````python
lista1 = [1, 2, 3]
lista1.append("hola")
print(lista1)
````

````python
lista1 = [1, 2, 3]
lista1.append(True)
print(lista1)
````

````python
lista1 = [1, 2, 3]
lista1.append(3.14)
print(lista1)
````

#### **8.2. `extend(iterable)`**
Añade cada elemento de un iterable (ej., lista, tupla) al final de la lista.

````python
lista1 = [1, 2, 3]
lista1.extend([4, 5, 6])
print(lista1)
````

````python
lista1 = [1, 2, 3]
lista1.extend("hola")
print(lista1)
````

````python
lista1 = [1, 2, 3]
lista1.extend(True)
print(lista1)
````

**Diferencia con `append()`**: `extend()` descompone el iterable en elementos individuales.

#### **8.3. `insert(índice, elemento)`**
Inserta un elemento en la posición especificada por `índice`.

````python
lista1 = [1, 2, 3]
lista1.insert(0, 0)
print(lista1)
````

````python
lista1 = [1, 2, 3]
lista1.insert(1, 10)
print(lista1)
````

````python
lista1 = [1, 2, 3]
lista1.insert(2, 20)
print(lista1)
````

 Si el índice es mayor que la longitud de la lista, el elemento se añade al final.


````python
lista1 = [1, 2, 3]
lista1.insert(10, 40)
print(lista1)
````

````python
lista1 = [1, 2, 3]
lista1.insert(3, 30)
print(lista1)
````

#### **8.4. `remove(elemento)`**
Elimina la primera aparición de `elemento` en la lista. Lanza `ValueError` si no se encuentra.

````python
lista1 = [1, 2, 3]
lista1.remove(2)
print(lista1)
````

````python
lista1 = [1, 2, 3]
lista1.remove(3)
print(lista1)
````

````python
lista1 = [1, 2, 3, 2]
lista1.remove(2)
print(lista1)
````

````python
lista1 = [1, 2, 3]
lista1.remove(4)
print(lista1)
````

#### **8.5. `pop([índice])`**
Elimina y devuelve el elemento en el `índice` especificado (por defecto, el último). Lanza `IndexError` si la lista está vacía o el índice es inválido.

````python
lista1 = [1, 2, 3]
print(lista1.pop())
print(lista1)
````

````python
lista1 = [1, 2, 3]
print(lista1.pop(0))
print(lista1)
````

````python
lista1 = [1, 2, 3]
print(lista1.pop(1))
print(lista1)
````

````python
lista1 = [1, 2, 3]
print(lista1.pop(2))
print(lista1)
````

````python
lista1 = [1, 2, 3]
print(lista1.pop(3))
print(lista1)
````

````python
lista1 = []
print(lista1.pop())
print(lista1)
````

#### **8.6. `clear()`**
Elimina todos los elementos de la lista, dejándola vacía.

````python
lista1 = [1, 2, 3]
lista1.clear()
print(lista1)
````

````python
lista1 = []
lista1.clear()
print(lista1)
````

#### **8.7. `index(elemento[, inicio[, fin]])`**
Devuelve el índice de la primera aparición de `elemento`. Lanza `ValueError` si no se encuentra.

````python
lista1 = [1, 2, 3]
print(lista1.index(2))
````

````python
lista1 = [1, 2, 3, 2]
print(lista1.index(2))
````

````python
lista1 = [1, 2, 3, 2]
print(lista1.index(6))
````

````python
lista1 = [1, 2, 3, 2]
print(lista1.index(2, 1))
````

````python
lista1 = [1, 2, 3, 2]
print(lista1.index(2, 2))
````

````python
lista1 = [1, 2, 3, 2]
print(lista1.index(2, 3))
````

````python
lista1 = [1, 2, 3, 2]
print(lista1.index(2, 0, 2))
````

````python
lista1 = [1, 2, 3, 2]
print(lista1.index(3, 0, 2))
````

#### **8.8. `count(elemento)`**
Cuenta cuántas veces aparece `elemento` en la lista.

````python
lista1 = [1, 2, 3]
print(lista1.count(2))
````

````python
lista1 = [1, 2, 3, 2]
print(lista1.count(2))
````

````python
lista1 = [1, 2, 3, 2]
print(lista1.count(6))
````

````python
lista1 = []
print(lista1.count(2))
````

#### **8.9. `sort(key=None, reverse=False)`**
Ordena la lista in situ. `key` define una función de ordenación personalizada; `reverse=True` ordena en orden descendente.

````python
lista1 = [3, 1, 2]
lista1.sort()
print(lista1)
````

````python
lista1 = [3, 1, 2]
lista1.sort(reverse=True)
print(lista1)
````

````python
lista1 = ["b", "a", "c"]
lista1.sort()
print(lista1)
````

````python
lista1 = ["b", "a", "c"]
lista1.sort(reverse=True)
print(lista1)
````

````python
lista1 = [3, 1, 2, "a", "b", "c"]
lista1.sort()
print(lista1)
````

Para obtener una copia ordenada sin modificar la lista original, usa `sorted(lista)`.

````python
lista1 = [3, 1, 2]
lista2 = sorted(lista1)
print(lista1)
print(lista2)
````

#### **8.10. `reverse()`**
Invierte el orden de los elementos de la lista in situ.

````python
lista1 = [1, 2, 3]
lista1.reverse()
print(lista1)
````

Equivalente a `lista[::-1]`, pero `reverse()` modifica la lista directamente.

````python
lista1 = [1, 2, 3]
lista1[::-1]
print(lista1)
````

#### **8.11. `copy()`**
Devuelve una copia superficial de la lista.

```python
lista = [1, 2, [3, 4]]
copia = lista.copy()
copia[0] = 99
print(lista)   # [1, 2, [3, 4]]
print(copia)   # [99, 2, [3, 4]]
copia[2][0] = 5  # Modifica la sublista en ambas listas
print(lista)   # [1, 2, [5, 4]]
print(copia)   # [99, 2, [5, 4]]
```

````python
lista = [1, 2, [3, 4]]
copia = lista.copy()
copia[0] = 99
print(lista)   # [1, 2, [3, 4]]
print(copia)   # [99, 2, [3, 4]]
copia[2][0] = 5  # Modifica la sublista en ambas listas
print(lista)   # [1, 2, [5, 4]]
print(copia)   # [99, 2, [5, 4]]
````
