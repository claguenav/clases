# **Strings**

### **1. ¿Qué es un string?**

Un **string** se corresponde con un conjunto de caracteres que forman una cadena de texto.

La sintaxis que debemos utilizar para definir strings en Python consiste en situar los caracteres entre `"` o `'`

````python
cadena1 = 'Hola, me llamo Gonzalo'
print(type(cadena1))
print(cadena1)
````

````python
cadena1 = "Hola, me llamo Gonzalo"
print(type(cadena1))
print(cadena1)
````

La flexibilidad de definir string con el caracter `"` y el caracter `'` nos permite definir cadenas de texto que contienen esos mismos caracteres.

````python
cadena1 = "Hola, me llamo 'Gonzalo'"
print(type(cadena1))
print(cadena1)
````

````python
cadena1 = 'Hola, me llamo "Gonzalo"'
print(type(cadena1))
print(cadena1)
````

### **2. Indexación**



En muchos tipos de datos en Python se puede acceder a elementos individuales de un conjunto ordenado de datos directamente mediante un índice numérico o un valor clave. Este proceso se denomina indexación.

En Python, las cadenas son secuencias ordenadas de caracteres, y por lo tanto pueden ser indexadas de esta manera. Se puede acceder a los caracteres individuales de una cadena especificando el nombre de la cadena seguido de un número entre corchetes `[]`.

El primer carácter de la cadena tiene el índice 0, el siguiente tiene el índice 1, y así sucesivamente. El índice del último carácter será la longitud de la cadena menos uno.

````python
texto = "Python"
print(texto[0])
print(texto[1])
print(texto[2])
print(texto[3])
print(texto[4])
print(texto[5])
````

Tambien podemos utilizar números negativos para extraer caracteres por el final de la cadena de texto.

````python
texto = "Python"
print(texto[-1])
print(texto[-2])
print(texto[-3])
print(texto[-4])
print(texto[-5])
print(texto[-6])
````

### **3. Slicing**

Python también permite una sintaxis específica de indexación que extrae subcadenas de una cadena de texto, a esto se denomina **slicing**.

La sintaxis que se utiliza para extraer una subcadena de una cadena `s` es de la forma `s[m:n]`, esto devuelve la porción de `s` que comienza en la posición `m`, y termina en la posición `n` (sin ser incluido el último caracter)

````python
texto = "Python"
print(texto[0:4])
print(texto[2:4])
print(texto[4:6])
````

````python
texto = "Python"
print(texto[0:15])
````

Si no indicamos uno de los números, lee hasta el final.

````python
texto = "Python"
print(texto[:2])
print(texto[2:])
````

### **4. Stride**

El stride es otra variante más del slicing. Si se añade un `:` adicional y un tercer índice, se designa una stride, que indica cuantos caracteres saltar hasta obtener el siguiente caracter.

````python
texto = "Python"
print(texto[0:6:2])
````

### **5. Modificación de strings**

Un string es un tipo de dato que Python considera **inmutable**, esto quiere decir que no podemos modificar una parte de un string asociada a una variable

````python
texto = "Jython"
texto[0] = "P"
````

A pesar de que no pueda modificarse el contenido de un string, si puede asignarse un string diferente a la variable.

````python
texto = "Jython"
texto = "Python"
print(texto)
````

````python
texto = "Jython"
texto = "P" + texto[1:]
print(texto)
````

### **6. Strings de múltiples líneas**

En algunas ocasiones es posible que queramos definir un string que tenga múltiples líneas. Existen varias formas de definir esto en Python.

La forma más sencilla es introducir el caracter `\n` en la posición de la cadena de texto donde queremos que se produzca el salto de línea.

````python
cadena1 = "Esta es la primera linea\nEsta es la segunda"
print(type(cadena1))
print(cadena1)
````

Otra opción interesante es situar nuestra cadena de texto entre los caracteres ```"""```.

````python
cadena1 = """Este es
un texto
multilinea"""
print(type(cadena1))
print(cadena1)
````

````python
cadena1 = '''Este es
un texto
multilinea'''
print(type(cadena1))
print(cadena1)
````

### **7. Operaciones Básicas con Cadenas**

- **Concatenación (`+`)**: Une cadenas.
- **Repetición (`*`)**: Repite una cadena.
- **Pertenencia (`[not] in`)**: Verifica si una subcadena está presente.
- **longitud (`len`)**: Indica el número de caracteres que forma la cadena de texto.

````python
cadena1 = "Hola" + " " + "amigo"
print(cadena1)
````

````python
cadena1 = "Hola"
cadena2 = "amigo"
cadena3 = cadena1 + cadena2
print(cadena3)
cadena3 = cadena1 + " " + cadena2
print(cadena3)
````

````python
texto = "Hola"
print(texto * 3)
````

````python
texto = "Hola"
print("la" in texto)
print("la" not in texto)
````

````python
texto = "Hola"
print(len(texto))
````

````python
texto = "Hola"
print(texto[0:len(texto)])
````

### **8. Métodos Comunes de Cadenas**

Python ofrece muchos métodos integrados para manipular cadenas. Aquí están los más útiles:

#### **8.1 Modificación de Contenido**
Estos métodos devuelven una nueva cadena, ya que las cadenas son **inmutables**.

- **`upper()`**: Convierte a mayúsculas.
- **`lower()`**: Convierte a minúsculas.
- **`capitalize()`**: Pone la primera letra en mayúscula.
- **`title()`**: Pone la primera letra de cada palabra en mayúscula.
- **`strip()`**: Elimina espacios en blanco (o caracteres específicos) al inicio y final.
- **`replace(viejo, nuevo)`**: Reemplaza una subcadena por otra.

````python
texto = "hola"
print(texto.upper())
````

````python
texto = "hola"
texto.upper()
print(texto)
````

````python
texto = "hola"
texto = texto.upper()
print(texto)
````

````python
texto = "HOLA"
print(texto.lower())
````

````python
texto = "hola"
print(texto.capitalize())
````

````python
texto = "esta frase tiene cinco palabras"
print(texto.title())
````

````python
texto = "      Esta cadena     tiene espacios      "
print(texto.strip())
````

````python
texto = "Hola mundo"
print(texto.replace("mundo", "Python"))
print(texto)
````

#### **8.2. Búsqueda y Verificación**
- **`find(sub)`**: Devuelve el índice de la primera aparición de `sub` o `-1` si no se encuentra.
- **`index(sub)`**: Similar a `find()`, pero lanza `ValueError` si no se encuentra.
- **`startswith(prefix)`**: Verifica si la cadena comienza con `prefix`.
- **`endswith(suffix)`**: Verifica si la cadena termina con `suffix`.
- **`count(sub)`**: Cuenta cuántas veces aparece `sub`.

````python
texto = "Esta es la cadena con el texto de lo que quiero decir"
print(texto.find("quiero"))
print(texto[texto.find("quiero"):])
````

````python
texto = "Esta es la cadena con el texto de lo que quiero decir"
print(texto[texto.find("quiero"):])
````

````python
texto = "Esta es la cadena con el texto de lo que quiero decir"
print(texto.find("hola"))
````

````python
texto = "Esta es la cadena con el texto de lo que quiero decir"
print(texto.index("quiero"))
````

````python
texto = "Esta es la cadena con el texto de lo que quiero decir"
print(texto.index("hola"))
````

````python
texto = "Esta es la cadena con el texto de lo que quiero decir"
print(texto.startswith("Esta"))
print(texto.startswith("hola"))
````

````python
texto = "Esta es la cadena con el texto de lo que quiero decir"
print(texto.endswith("decir"))
print(texto.endswith("hola"))
````

````python
texto = "Esta es la cadena con el texto de lo que quiero decir"
print(texto.count("qu"))
````

#### **8.3. Validación de Contenido**
- **`isalpha()`**: Verdadero si todos los caracteres son letras.
- **`isdigit()`**: Verdadero si todos los caracteres son dígitos.
- **`isalnum()`**: Verdadero si todos los caracteres son alfanuméricos.
- **`isspace()`**: Verdadero si todos los caracteres son espacios.

````python
texto = "1000"
print(texto.isalpha())
````

````python
texto = "mil"
print(texto.isalpha())
````

````python
texto = "Te he escrito un cheque por mil euros"
print(texto.isalpha())
````

````python
texto = "1000"
print(texto.isdigit())
````

````python
texto = "P4ssw0rd"
print(texto.isalnum())
````

````python
texto = "        "
print(texto.isspace())
````

#### **8.4. División y Unión**
- **`split(sep)`**: Divide la cadena en una lista usando `sep` (por defecto, espacio).
- **`join(iterable)`**: Une elementos de un iterable en una cadena, usando la cadena como separador.

````python
texto = "Hola mundo Python"
palabras = texto.split()
print(palabras)
````

````python
texto = "Hola-mundo-Python"
palabras = texto.split("-")
print(palabras)
````

````python
texto = "las amigas lamentaron la pérdida de la perla negra"
palabras = texto.split("la")
print(palabras)
````

````python
texto = "este es el texto que quiero unir con guiones"
palabras = texto.split()
nuevo_texto = "-".join(palabras)
print(nuevo_texto)
````
